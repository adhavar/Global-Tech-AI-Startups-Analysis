## Loading Libraries

```python
import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
from statsmodels.stats.anova import anova_lm as anova
```

## Research Question
**Is the AI adoption level of global tech and AI startups associated with stronger financial performance and better operational metrics?**

## Null and Alternative Hypotheses for ANOVA
**H₀**: The mean of the performance variable is equal across all AI adoption levels.

**H₁**: At least one AI adoption level has a different mean for the performance variable.

## Why ANOVA?

ANOVA (Analysis of Variance) is a statistical test used to determine the differences in the mean of a quantitative variable across three or more groups. In this case, I need to compare quantitative variables (the financial performance and operational metrics) across four AI adoption levels (No AI adoption, Low, Medium and High). Since the mean is a suitable statistical measure to compare the performance of each variable across the different levels, I decided to use ANOVA.

A one-way ANOVA will be used to analyze each quantitative variable. If the difference for the variable *x* is statistically significant, I will have evidence to suggest that there is a difference in performance among the various levels of AI adoption, so I will identify which levels differ from each other using Tukey’s HSD test. If there is no statistically significant difference, then I will not have sufficient evidence to suggest that the level of AI adoption is associated with the startups' performance.

## Loading the clean dataset

```python
data = pd.read_csv("global_tech_startups_cleaned.csv")
```

## Creating the key variables

```python
# Selection of the quantitative key variables
performance_variables = [
    'Revenue_ARR_Millions',
    'Valuation_USD_Millions',
    'Total_Funding_USD_Millions',
    'Layoffs_2024_2025',
    'Current_Headcount_2026',
    'Runway_Months_2024'
]
```

```python
# for each variable, the model and the if/else statements will be tested
for variable in performance_variables:
    # ANOVA model
    model = ols(f'{variable} ~ C(AI_Adoption_Level)', data = data).fit()

    # ANOVA table
    anova_table = anova(model, typ = 2)

    print(variable)
    display(anova_table)

    # Identification of PR(>F)
    p_value = anova_table.loc['C(AI_Adoption_Level)', 'PR(>F)']

    # if/else statements to determine if H₀ is rejected or not
    if p_value < 0.05:
        print('Reject H₀' + '\n' + '__________________________________________________________')
    else:
        print('Fail to reject H₀' + '\n' + '__________________________________________________________')

    print()
```

## Interpretation

Since the p-values for all the performance variables are greater than the significance level (α = 0.05), there is not enough evidence to reject the null hypothesis. None of the variables show a statistically significant difference across AI adoption levels.

Therefore, there is not enough evidence to believe that the AI adoption level of global tech and AI startups is associated with their performance.

Because there were no significant differences, Tukey HSD was not necessary.
